# A. WORD PROBLEM ROTATIONS

```python
def rotate_word(text):
    # making the first character go to the end
    return text[1:] + text[0]


# input a word
word = input("Enter a word: ")

# Checking if it entered a word
if word != "":
    # Display the rotated word
    print("Rotated word:", rotate_word(word))
else:
    # Display a message if the user did not enter anything
    print("Please enter a word!!!!!!")
```

# B. USERNAME BUILDER PROBLEM

```python
def make_username(first_name, last_name):
    #Converting it to lower case
    first_name = first_name.lower()
    last_name = last_name.lower()

    #Remove spaces
    first_name = first_name.replace(" ", "")
    last_name = last_name.replace(" ", "")

    #Combining the two names
    user = first_name + "." + last_name

    return user


# Ask the user to enter their name
first_name = input("Enter your first name: ")
last_name = input("Enter your last name: ")

# Display
print("Your username is:", make_username(first_name, last_name))
```

# C. BOOKEND SWAP PROBLEM

```python
def swap_bookends(word):
    # Separate the first, middle, and last word
    first, *middle, last = word

    # Swap the first and last word
    new = [last] + middle + [first]

    return new


# Ask the user to enter text
text = input("Enter words separated by spaces: ")

# Convert the text into a list
word = text.split()

# Check if there are at least two words
if len(word) >= 2:
    result = swap_bookends(word)

    # Display the result as text
    print("Swapped text:", " ".join(result))
else:
    print("Please enter at least two words.")
```
